### Homework Hack Predictions & Explanation

**Calculations with Quantity = 4:**
* **Subtotal:** 12 * 4 = **48**
* **Discount:** 48 * 0.25 = **12**
* **Total:** 48 - 12 = **36**
* **Within Budget:** 36 <= 30 evaluates to **false**

**Explanation:**
When the quantity increases from 3 to 4, the subtotal jumps to 48. Even with a 25% discount saving 12, the final total comes out to 36. Because 36 is greater than our budget of 30, the `withinBudget` comparison evaluates to `false`.
